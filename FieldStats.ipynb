### Count values in a field

**Define the field (**`field_name`**) and target value (**`target_value`**) to be counted**

```python
field_name_count = "outcome"
target_value = "Preferred site (to meet SS)"
```

```python
count = len(
    [
        row[0]
        for row in arcpy.da.SearchCursor(feature_class, [field_name_count])
        if row[0] == target_value
    ]
)

print(f"Number of '{target_value}' sites: {count}")
```

### Get a total sum for values in a field

**Define the field name and feature layer**

```python
field_name_sum = "years1to10units"
```

**Sum the field**

if `sum` is overwritten somewhere else in the script, run `del sum` below to fix this

```python
# del sum
```

```python
# Extract values, filtering out None (null) values, then sum
total = sum(
    row[0]
    for row in arcpy.da.SearchCursor(feature_class, [field_name_sum])
    if row[0] is not None
)
print(f"Total units: {total}")

# 7819 when including pending sites, otherwise 7409
```
