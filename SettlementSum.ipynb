### Sum and count of settlements with no primary school (not within 1km)

```python
import pandas as pd
```

```python
df = pd.read_excel(r"S:\data\Devment\ArcGISPro\mneuwirth\GIS Projects (SDrive)\LDP3_Mapping\spreadsheets\Development_In_settlements_away_from_primary_schools_TableToExcel.xlsx", usecols=[5, 7, 9])# set these to the column indices of the settlement and units 
print(df)
```

**Get count of allocations per settlement**

```python
df_count = df.groupby(['settlement']).allocation.count().reset_index()
print(df_count)
```

**Get sum of units per settlements**

```python
df_sum = df.groupby(['settlement']).years1to10units.sum().reset_index()
print(df_sum.sort_values('years1to10units', ascending=False))
```

**Merge data frames for export**

```python
sum_count_df = pd.merge(df_count, df_sum, on='settlement', how='inner')
print(sum_count_df)
```

**Export to excel sheet**

```python
sum_count_df.to_excel('spreadsheets\outputs\settlements_no_primary_school.xlsx')
```

### Sum and count of settlements with primary schools

```python
import pandas as pd
```

```python
df = pd.read_excel("S:\data\Devment\ArcGISPro\mneuwirth\GIS Projects (SDrive)\LDP3_Mapping\spreadsheets\Development_in_Settlements_W_PS_TableToExcel.xlsx", usecols=[6, 8, 5])# set these to the column indices of the settlement and units 
print(df)
```

```python
unit_sum = df['years1to10units'].sum()
print(unit_sum)
```

```python
df_count = df.groupby(['settlement']).allocation.count().reset_index()
print(df_count)
```

```python
df_sum = df.groupby(['settlement']).years1to10units.sum().reset_index()
print(df_sum.sort_values('years1to10units', ascending=False))
```

```python
sum_count_df = pd.merge(df_count, df_sum, on='settlement', how='inner')
print(sum_count_df)
```

```python
sum_count_df.to_excel('spreadsheets\outputs\settlements_WITH_primary_school.xlsx')
```

```python
import matplotlib.pyplot as plt
```

```python
sorted_df = sum_count_df.sort_values('years1to10units', ascending=True)
```

```python
sorted_df.plot(x='settlement', y='years1to10units', kind='barh', figsize=(8,10), label='Units (1-10 years)')
plt.show
```
