## Excel column to SQL query

Script to extract data from excel column and print it into the fromat required for SQL qeuries. 

```python
import pandas as pd
```

```python
# Path to excel file and define column name and sheet number
# Use filter column in the spreadsheet located on the Sdrive, not in EDMS
file_path = "S:\data\Devment\ArcGISPro\mneuwirth\GIS Projects (SDrive)\LDP3_Mapping\SptatialStratagyLayerJoin.xlsx"
column_name = 'Filter'
sheet_name = 4  # Sheet index or sheet name (put 0 for sheet 1)

# Read the Excel file
df = pd.read_excel(file_path, sheet_name=sheet_name)

# Drop missing values and convert all items to strings
data = df[column_name].dropna().astype(str).tolist()

# Format each value with single quotes and join with commas
result = ", ".join(f"'{item}'" for item in data)

# Output the formatted string
print(f"siteid IN ({result})")
```

#### Housing units

```python
file_path = r"S:\data\Devment\ArcGISPro\mneuwirth\GIS Projects (SDrive)\LDP3_Mapping\LDP3_Spatial_Strategy_Dashboard_Join_V4_Excel_Table.xlsx"
sheet_number = 0

df = pd.read_excel(file_path, sheet_name=sheet_number)

gis_dict = dict(zip(df.iloc[:, 2], df.iloc[:, 3]))

for _, row in df.iterrows():
    site_id = row.iloc[0]
    housing_1 = row.iloc[1]

    if site_id in gis_dict:
        housing_2 = gis_dict[site_id]

        if housing_1 != housing_2:
            print(
                f"Site ID: {site_id} | "
                f"Spreadsheet: {housing_1} | "
                f"GIS: {housing_2}"
            )
```
